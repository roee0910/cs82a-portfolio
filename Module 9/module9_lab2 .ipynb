# Migration Note: Google Colab to Kaggle Notebooks

**To:** Course team  |  **From:** Roey Kadur |  **Subject:** Moving the Module 9 course workflow  
**Decision:** Kaggle Notebooks is a workable alternative for running this Python notebook, after a short setup and file check.

## What transfers untouched

- **Notebook structure:** The `.ipynb` file and its Markdown/code cells can be uploaded to Kaggle and opened as a notebook.
- **Python code:** Standard Python and common data-science code should transfer without changes. Code that depends on Colab-only helpers (for example, `google.colab.files` or `drive.mount`) must be replaced with Kaggle-compatible file access.
- **Version history copy:** Keep the original notebook in the portfolio repository; upload a working copy to Kaggle so the submitted source remains reviewable.

## What needs to be redone

| Item | Action on Kaggle |
|---|---|
| Account and access | Sign in or create a Kaggle account, accept applicable platform terms, and confirm notebook access before the deadline. |
| Notebook and storage | Upload the `.ipynb`; add input files through Kaggle's notebook data/input workflow or another approved source. Colab Drive mounts and local runtime paths do not carry over. |
| Packages | Inspect imports, then install missing non-default packages in the Kaggle session (or select an environment that provides them). Record versions for packages that affect results. |
| Runtime settings | Select an available accelerator only if the work needs it; recheck any hardware-specific assumptions and rerun all cells from the top. |
| Outputs and links | Regenerate outputs in the new session, save required files to a persistent/downloadable location, and update any Colab-specific links in the notebook. |

## Risk and mitigation

**Risk:** A notebook can open successfully while still depending on a file stored only in a Colab session or on a package installed only in that session. The result may fail partway through or produce different outputs after migration.  
**Mitigation:** Keep inputs in a named, documented location; add a setup cell for required packages; record package versions; then restart the Kaggle session and run every cell top to bottom before treating the migration as complete.

## Experience from the local Jupyter to Colab move

In Module 1 Lab 2, my notebook loaded `chickwts.csv` with `pd.read_csv("chickwts.csv", index_col=0)`, using a relative file path. That taught me to check the data file location when moving notebooks between platforms: the notebook can transfer while its input file still needs to be added to the new platform and made available at the expected path. In this migration, I would verify that input step before running the analysis.
