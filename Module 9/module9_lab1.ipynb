## Problem
Write and demonstrate a function that converts pounds to kilograms (1 lb = 0.453592 kg), and call it on at least three values, one of which is my own weight. The entire workflow runs in Google Colab, with nothing installed on my own machine.

## 1. The conversion function

```python
# Conversion factor: 1 pound = 0.453592 kilograms
KG_PER_LB = 0.453592

def pounds_to_kilograms(pounds):
    """Convert a weight in pounds to kilograms."""
    return pounds * KG_PER_LB

print("Function defined: pounds_to_kilograms(pounds)")
```

## 3. Three calls to the function
The third value is my own weight. 

```python
# Conversion factor: 1 pound = 0.453592 kilograms
KG_PER_LB = 0.453592

def pounds_to_kilograms(pounds):
    """Convert a weight in pounds to kilograms."""
    return pounds * KG_PER_LB

my_weight_kg = 70                         # my weight in kg
my_weight_lb = my_weight_kg / KG_PER_LB   # about 154.32 lb

test_values = [10, 150, my_weight_lb]
labels = ["10 lb (test value)", "150 lb (test value)", "My own weight"]

for label, lb in zip(labels, test_values):
    kg = pounds_to_kilograms(lb)
    print(f"{label:22s}: {lb:7.2f} lb = {kg:8.3f} kg")
```

## 4. Checking the results
Hand calculation: 10 lb x 0.453592 = 4.53592 kg, and 150 lb x 0.453592 = 68.0388 kg. The assertions below raise an error if the function is wrong.

```python
assert abs(pounds_to_kilograms(10) - 4.53592) < 1e-9
assert abs(pounds_to_kilograms(150) - 68.0388) < 1e-9
assert pounds_to_kilograms(0) == 0
print("All checks passed.")
```
